## 3. Structural Conditioning with ControlNet Canny
Extracting Canny edge maps locks the room geometry, bed borders, and pillars so Gen AI doesn't distort property features.

```python
import cv2

def extract_canny(image_pil):
    img_np = np.array(image_pil)
    gray = cv2.cvtColor(img_np, cv2.COLOR_RGB2GRAY)
    edges = cv2.Canny(gray, 100, 200)
    return Image.fromarray(cv2.cvtColor(edges, cv2.COLOR_GRAY2RGB))

print("Canny extraction helper defined.")
```
